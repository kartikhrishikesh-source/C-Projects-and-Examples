```python
%%bash
gcc -o hello_c hello_c.c
./hello_c

cat <<EOF > hello_c.c
#include <stdio.h>

int main() {
    char name[] = "John";
    printf("Hello %s",name);
    return 0;
}
EOF
```

```python
%%bash
cat <<EOF > hello.c
#include <stdio.h>

int main() {
    printf("Hello World!\n\n");
    printf("I like programming in C");
    return 0;
}
EOF

gcc -o hello hello.c
./hello
```

```python
%%bash
cat <<EOF > ben.c
#include <stdio.h>
int main() {
  printf("They call him \"Johnny\".");
}
EOF

gcc -o ben ben.c
./ben
```
