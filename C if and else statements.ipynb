```python
%%bash

echo '#include <stdio.h>
#include <stdbool.h>

int main() {
  int x = 20;
  int y = 18;

  bool isGreater = x > y;

  if (isGreater) {
    printf("X is greater than Y");
  }
  return 0;
}' > main.c


gcc main.c -o main


./main
```
