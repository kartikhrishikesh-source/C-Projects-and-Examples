```python
%%bash
gcc -o my_program -xc - <<'EOF'
#include <stdio.h>

int main() {
  int myNum;
  printf("Type a number and press enter: ");
  scanf("%d", &myNum);
  printf("Your number is: %d\n", myNum);
  return 0;
}
EOF
./my_program
```

```python
%%bash
gcc -o my_program -xc - <<'EOF'
#include <stdio.h>
int main() {
  int myNum;
  char myChar;

  printf("Type a number and press enter: \n");
  scanf("%d %c", &myNum, &myChar);

  printf("Your number is: %d\n", myNum);
  printf("Your character is: %c\n", myChar);

  return 0;

}
EOF
./my_program
```

```python
%%bash
gcc -o my_program -xc - <<'EOF'
#include <stdio.h>

int main() {
  char firstName[30];

  printf("Enter your first name and press enter: \n");
  scanf("%s", firstName);
  printf("Hello %s\n", firstName);
  return 0;
}
EOF
./my_program
```

```python
%%bash
gcc -o my_program -xc - <<'EOF'
#include <stdio.h>

int main() {
  char fullName[30];
  printf("Type your full name and press enter: \n");
  fgets(fullName, sizeof(fullName), stdin);
  printf("Hello %s", fullName);
  return 0;
}
EOF
./my_program
```
