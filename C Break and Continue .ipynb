```python
%%bash
cat << EOF > main.c
#include <stdio.h>

int main() {
  int i = 0;

  while (i < 10) {
    if (i == 4) {
      i++;
      continue;
    }
    printf("%d\n", i);
    i++;
  }
  return 0;
}
EOF

gcc main.c -o main
./main
```

```python
%%bash
cat << EOF > main.c
#include <stdio.h>

int main() {
  int numbers[] = {3, -1, 7, 0, 9};
  int length = sizeof(numbers) / sizeof(numbers[0]);
  int i;

  for (i = 0; i < length; i++) {
    if (numbers[i] < 0) {
      continue;
    }
    if (numbers[i] == 0) {
      break;
    }
    printf("%d\n", numbers[i]);
  }
  return 0;
}
EOF

gcc main.c -o main
./main
```
