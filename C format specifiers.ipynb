```python
%%writefile program.c
#include <stdio.h>

int main() {
  int myNum = 15;
  float myFloatNum = 19.99;
  char myLetter = 'D';

  printf("%d\n", myNum);
  printf("%f\n", myFloatNum);
  printf("%c\n", myLetter);
  return 0;
}

%%bash
gcc program.c -o program
./program
```

```python
%%writefile sum_program.c
#include <stdio.h>

int main() {
  int x, y, z;
  x = y = z = 50;
  printf("%d", x + y + z);
  return 0;
}

%%bash
gcc sum_program.c -o sum_program
./sum_program
```

```python
%%writefile area_calculator.c
#include <stdio.h>

int main() {
  int length = 4;
  int width = 6;
  int area;

  area = length * width;

  printf("The length is: %d\n", length);
  printf("The width is: %d\n", width);
  printf("The area is: %d\n", area);
  return 0;
}

%%bash
gcc area_calculator.c -o area_calculator
./area_calculator
```
