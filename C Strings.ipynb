```python
%%bash
cat > car_name_length.c <<EOF
#include <stdio.h>

int main() {
  char carName[] = "Volvo";


  int length = sizeof(carName) / sizeof(carName[0]) - 1; // Subtract 1 for the null terminator
  int i;

  for (i = 0; i < length; ++i) {
    printf("%c\n", carName[i]);
  }
  return 0;
}
EOF
gcc -o car_name_length car_name_length.c
./car_name_length
```

```python
%%bash
cat > string_concat.c <<EOF
#include <stdio.h>
#include <string.h>

int main() {
  char str1[20] = "Hello "; // Added space for better concatenation
  char str2[] = "World";

  strcat(str1, str2);
  printf("%s\n", str1);
  return 0;
}
EOF
gcc -o string_concat string_concat.c
./string_concat
```

```python
%%bash
cat > string_copy.c <<EOF
#include <stdio.h>
#include <string.h>

int main() {
  char str1[20] = "Hello World";
  char str2[20];

  strcpy(str2, str1);
  printf("%s\n", str2);
  return 0;
}
EOF
gcc -o string_copy string_copy.c
./string_copy
```

```python
%%bash
cat > string_compare.c <<EOF
#include <stdio.h>
#include <string.h>

int main() {
   char str1[] = "Hello";
   char str2[] = "Hello";
   char str3[] = "Hi";

   printf("%d\n", strcmp(str1, str2));
   printf("%d\n", strcmp(str1, str3));

  return 0;
}
EOF
gcc -o string_compare string_compare.c
./string_compare
```
