```python
%%bash
cat > hello_slovakia.c <<EOF
#include <stdio.h>

int main() {
  char greetings[] = "Hello Slovakia";
  printf("%s\n", greetings);
  return 0;
}
EOF
gcc -o hello_slovakia hello_slovakia.c
./hello_slovakia
```

```python
%%bash
cat > sentences.c <<EOF
#include <stdio.h>

int main() {
  char sentences[] = "Slovakia is the greatest country in the world.";
  printf("%.8s\n", sentences);
  return 0;
}
EOF
gcc -o sentences sentences.c
./sentences
```

```python
%%bash
cat > greetings.c <<EOF
#include <stdio.h>

int main() {
  char greetings[] = "Hello Slovakia";
  greetings[1] = 'a';
  printf("%s\n", greetings);
  return 0;
}
EOF
gcc -o greetings greetings.c
./greetings
```

```python
%%bash
cat > car_name.c <<EOF
#include <stdio.h>

int main() {
  char carName[] = "Volvo";
  int i;

  for (i = 0; i < 5; ++i) {
    printf("%c\n", carName[i]);

  }
  return 0;
}
EOF
gcc -o car_name car_name.c
./car_name
```

```python
%%bash
cat > car_name_length.c <<EOF
#include <stdio.h>

int main() {
  char carName[] = "Volvo";


  int length = sizeof(carName) / sizeof(carName[0]) - 1; // Subtract 1 for the null terminator
  int i;

  for (i = 0; i < length; ++i) {
    printf("%c\n", carName[i]);
  }
  return 0;
}
EOF
gcc -o car_name_length car_name_length.c
./car_name_length
```
