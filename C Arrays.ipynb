```python
%%bash

cat << EOF > program.c
#include <stdio.h>

int main() {
  int myNumbers[] = {25, 50, 75, 100};
  printf("%d", myNumbers[0]);
  return 0;
}
EOF

gcc program.c -o program


./program
```

```python
%%bash
cat << EOF > program.c
#include <stdio.h>

int main() {
  float myArray[] = {25, 50, 75, 3.15, 5.99};
  int i;

  for (i = 0; i < 5; i++) {
    printf("%f\n", myArray[i]);
  }
  return 0;
}
EOF

gcc program.c -o program

./program
```
