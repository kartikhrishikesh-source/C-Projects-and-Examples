```python
%%bash
cat << 'EOF' > division.c
#include <stdio.h>
int main() {
  int x = 10;
  int y = 0;
  printf("Before division\n");

  if (y != 0) {
    int z = x / y;
    printf("Result: %d\n", z);
  } else {
    printf("Error: Divison by zero!\n");
  }
  printf("After division\n");
  return 0;
}
EOF
gcc division.c -o division
./division
```

```python
%%bash
cat << 'EOF' > bounds_check.c
#include <stdio.h>
int main() {
  int numbers[3] = {10,20,30};
  int index = 5;

  printf("Index = %d\n",index);
  if (index >= 0 && index < 3) {
    printf("Value = %d\n", numbers[index]);
  } else {
    printf("Error: Index out of bounds!\n");
  }
  return 0;
}
EOF
gcc bounds_check.c -o bounds_check
./bounds_check
```

```python
%%bash
cat << 'EOF' > file_check.c
#include <stdio.h>
int main() {
  FILE *fptr = fopen("nothing.txt", "r");
  if (fptr == NULL) {
    printf("Error: File not found!\n");
    return 1;
  }
  fclose(fptr);
  return 0;
}
EOF
gcc file_check.c -o file_check
./file_check || true
```

```python
%%bash
cat << 'EOF' > memory_check.c
#include <stdio.h>
#include <stdlib.h>

int main() {
  int *numbers = malloc(100000000000 * sizeof(int));
  if (numbers == NULL) {
    printf("Memory allocation failed.\n");
    return 1;
  } else {
    printf("Memory allocation successful.\n");
    free(numbers);
    numbers = NULL;
    return 0;
  }
}
EOF
gcc memory_check.c -o memory_check
./memory_check || true
```

```python
%%bash
cat << 'EOF' > divide.c
#include <stdio.h>

int divide(int a, int b, int *result) {
  if (b == 0) {
    return -1;
  }
  *result = a / b;
  return 0;
}

int main() {
  int result;
  if (divide(10, 0, &result) != 0) {
    printf("Error: Divison by zero!\n");
    return 1;
  }
  printf("Result: %d\n", result);
  return 0;
}
EOF
gcc divide.c -o divide
./divide || true
```

```python
%%bash
cat << 'EOF' > errno_check.c
#include <stdio.h>
#include <string.h>
#include <errno.h>

int main() {
  FILE *f = fopen("nothing.txt","r");
  if (f == NULL) {
    printf("Error: %s\n", strerror(errno));
    return 1;
  }
  fclose(f);
  return 0;
}
EOF
gcc errno_check.c -o errno_check
./errno_check || true
```

```python
%%bash
cat << 'EOF' > errno_test.c
#include <stdio.h>
#include <stdlib.h>
#include <errno.h>

int main() {
  FILE *f = fopen("nothing.txt", "r");
  if ( f == NULL) {
    if (errno == ENOENT) {
      printf("The file was not found.\n");
    } else {
      printf("Some other file error occured.\n");
    }
    return 1;
  }
  fclose(f);
  return 0;
}
EOF
gcc errno_test.c -o errno_test
./errno_test || true
```

```python
%%bash
cat << 'EOF' > file_check_exit.c
#include <stdio.h>
#include <stdlib.h>

int main() {
  FILE *f = fopen("nothing.txt", "r");
  if (f == NULL) {
    printf("Failed to open file.\n");
    exit(1);
  }
  fclose(f);
  return 0;
}
EOF
gcc file_check_exit.c -o file_check_exit
./file_check_exit || true
```

```python
%%bash
cat << 'EOF' > program_exit.c
#include <stdio.h>
#include <stdlib.h>

int main() {
  FILE *f = fopen("nothing.txt", "r");
  if (f == NULL) {
    perror("Could not open nothing.txt");
    exit(EXIT_FAILURE);
  }
  fclose(f);
  return EXIT_SUCCESS;
}
EOF
gcc program_exit.c -o program_exit
./program_exit || true
```
